```python
#TYPE OF TRIANGLE

def triangle():
    s1=int(input("Enter the side of your triangle"))
    s2=int(input("Enter the side of your triangle"))
    s3=int(input("Enter the side of your triangle"))
    if (s1==s2==s3):
        print("It's an eu=quilateral triangle")
    elif (s1==s2 or s2==s3 or s1==s3):
        print("It's an isosceles triangle")
    else :
        print("It's a scalene triangle")

triangle()
```
